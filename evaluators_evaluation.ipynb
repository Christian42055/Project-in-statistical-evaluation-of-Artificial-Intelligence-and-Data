Code for the evaluation of the evaluator LLMs (oss 20b, oss 120b and llama4scout)

```python
import pandas as pd
import numpy as np
```

```python
eval_files = [
    r"evaluators_evaluator_csv/evaluation_test_20b.csv",
    r"evaluators_evaluator_csv/evaluation_test_120b.csv",
    r"evaluators_evaluator_csv/evaluation_test_scout.csv"
]
```

```python
results = []
questions = ["q1", "q2", "q3", "q4", "q5", "q6"]
models = ["20b", "120b", "scout"]

for q in questions:
    row = {"question": q}

    for n, file in enumerate(eval_files):
        df = pd.read_csv(file)
        df = df[questions].apply(pd.to_numeric, errors="coerce")

        barcelona_mean = df.head(20).tail(20)[q].mean()
        groceries_mean = df.head(40).tail(20)[q].mean()
        tmdb_mean = df.head(60).tail(20)[q].mean()
        overall_mean = df[q].mean()

        row[f"mean barcelona {models[n]}"] = barcelona_mean
        row[f"mean groceries {models[n]}"] = groceries_mean
        row[f"mean TMDB {models[n]}"] = tmdb_mean
        row[f"overall mean {models[n]}"] = overall_mean

    results.append(row)

results_df = pd.DataFrame(results)
results_df = results_df.round(decimals=5)
results_df.to_csv(r"finished_work/pilot_run.csv", index=False)

results_df
```
