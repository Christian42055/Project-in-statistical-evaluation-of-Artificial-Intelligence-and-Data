```python
import pandas as pd
```

```python
# df1 = pd.read_csv(r"csv_files/explainDepression_final_dataset.csv")
df2 = pd.read_csv(r"csv_files/mental_health_combined_test.csv")
df3 = pd.read_csv(r"csv_files/Mental_Health_Condition_Classification.csv")
df4 = pd.read_csv(r"csv_files/mental_heath_feature_engineered.csv")
df5 = pd.read_csv(r"csv_files/mental_heath_unbanlanced.csv")
df6 = pd.read_csv(r"csv_files/Sentiment_Mental_health_dataset.csv")
```

```python
texts = pd.concat(
    [
        # df1["text"],
        df2["text"],
        df3["text"],
        df4["text"],
        df5["text"],
        df6["statement"],
    ],
    ignore_index=True
)

new_df = pd.DataFrame({"text": texts})

new_df = new_df.dropna(subset=["text"])
new_df = new_df[new_df["text"].str.strip() != ""]

new_df.to_csv("stories.csv", index=False)
```
