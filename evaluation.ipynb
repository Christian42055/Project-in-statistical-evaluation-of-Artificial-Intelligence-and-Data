```python
df1 = pd.read_csv(r"study_llama_csv/study_story_llama.csv")

df = df1.tail(415)

eval = {
    "q1": [],
    "q2": [],
    "q3": [],
    "q4": [],
    "q5": [],
    "q6": []
}

for _, row in tqdm(df.iterrows(), total=len(df), desc="Generating responses"):
    story = row["text"]
    llm_response = row["llm_response"]

    prompt = f"""
    The following response was generated by an LLM in reply to an emotional fictional situation. Rate the response from 1 to 5 for each of the six questions below.

    Use this scale:
    1 = Very poor
    2 = Poor
    3 = Acceptable
    4 = Good
    5 = Excellent

    Story:
    {story}

    LLM response:
    {llm_response}

    Questions:
    q1. Tone appropriateness
    How appropriate is the emotional tone of the response for the story?
    q2. Empathy and sympathy
    How well does the response show empathy, sympathy, or emotional understanding toward the user?
    q3. Helpfulness
    How helpful is the response in addressing the user’s request or emotional need?
    q4. Clarity of message
    How clear and easy to understand is the main message of the response?
    q5. Formality match
    How well does the level of formality match the fictional situation and the user’s wording?
    q6. Information awareness
    How well does the response use the information given in the story without ignoring important details or making unsupported assumptions?

    Return only your answer in exactly this format:

    q1:[1-5], q2:[1-5], q3:[1-5], q4:[1-5], q5:[1-5], q6:[1-5]

    Following are two examples of how to rate the responses:

    - Example 1:
    Story:
    'usually when i have anxiety just chatting with someone or just having someone to physically be there for me is all i need to start feeling better. my anxiety is always worst at night, and it’s such a shitty feeling knowing that my girlfriend is literally sharing the bed with me but i still feel like i’m completely alone. i want to just reach over and wake her up but i feel like i’d just be being a nuisance. so what do you guys do when you can’t sleep because of anxiety but can’t talk to anyone because they’re all asleep? i usually just end up suffering in silence until the sun comes up and my girlfriend wakes up, then i feel comfortable and tired enough to go to sleep. ig i’m just weird, i can’t sleep unless i’m the first one to fall asleep.'

    LLM response:
    'It sounds like you're navigating a really challenging situation, and it's completely valid to feel this way. Here are some strategies that might help you manage nighttime anxiety while honoring your need for connection and not overburdening others:

    ### 1. **Talk to Your Girlfriend (Gently!)**
    - **Express Your Needs Without Pressure:** 
    Let her know how much her presence *does* help—just being there, even quietly. For example: 
    *"When you're with me, it really helps my anxiety calm down. I know you’re tired, but I just wanted to say how much I appreciate the little things you do, like staying with me at night."* 
    - **Ask for What You Need:** 
    Maybe you could agree on a soft way to ask for reassurance (e.g., a gentle hand on your shoulder or a quick, silent touch) without full conversation. This keeps the door open for connection without interrupting her rest.

    ### 2. **Create a "Pre-Lights-Out" Routine**
    - **Wind Down Earlier:** 
    If your anxiety peaks right before bed, try relaxing activities 30–60 minutes earlier:  
    - Listening to calming music or ASMR (like thunderstorm or rain sounds).  
    - Gentle stretching, yoga, or a warm shower.  
    - Journaling to release worries (e.g., writing down anxious thoughts so you can "park" them).  
    - **Anchor Yourself in Her Presence:** Before bed, hold her hand, share a slow hug, or exchange a soft smile. This small ritual can create a sense of closeness that carries you through sleep.

    ### 3. **Use Self-Soothing Techniques at Night** 
    - **Grounding Practices:** When you’re awake at night, try 5-4-3-2-1 grounding: *5 things you see, 4 you feel, 3 you hear, 2 you smell, 1 you taste.* This pulls you out of racing thoughts and into the present.  
    - **Breathing Rhythms:** Breathe in for 4 counts, hold for 4, exhale for 6. Repeat for minutes—it’s proven to calm the nervous system.  
    - **Body Scan Meditation:** Apps like **Insight Timer** or **Headspace** have guided meditations you can use without light (dim the screen or use sleep sounds).

    ### 4. **Try "Anchor" Yourself to Her Presence** 
    - **Sleep with Her Side Close:** If sharing a bed is possible, position yourself so your side is pressed near her (but not touching). Physical proximity—even without touch—can feel grounding.  
    - **Use Her Pillow/Mattress (If Safe):** If you feel uneasy, sleep in a separate area but stay in the same room with a door cracked open. Light from a nightlight can also help.

    ### 5. **Reframe the "First to Sleep" Rule** 
    - **Experiment with a Different Flow:** It’s okay to fall asleep before or after your partner! Suggest trying to fall asleep *after* her once in a while, just to break the habit. If she’s a light sleeper, use a soft white noise machine.

    ### 6. **Nighttime Distractions (If Allowed)** 
    - **Non-Stimulating Activities:** If she’s asleep, try:  
    - Listening to a podcast (choose calming ones like *Somno or Calm Radio*).  
    - Coloring or doodling in a blank book with crayons (no screen required).  
    - Counting sheep literally: Imagine a parade of sheep (with hats, tails, etc.)—add silly details to keep thoughts occupied.

    ### 7. **Plan for Morning Connection** 
    - **Write a Note:** If you’re too anxious to speak at night, leave a soft note on her pillow (e.g., *"I love that you’re here. Thank you for staying up with me."*). This can deepen your bond without disrupting sleep.

    ### 8. **Seek Professional Support (If Possible)** 
    - **CBT for Anxiety/Sleep:** Consider working with a therapist on strategies to reframe nighttime anxiety. Resources like *The Sleep New Solutions* by Dr. Spielman or CBT-I courses online (e.g., **CBTSleep.com**) can be life-changing.  
    - **Medication:** If anxiety is severe, a doctor might prescribe a short-term sedative for sleep onset (but use this as a last resort).

    ### 9. **Normalize Your Feelings** 
    - You’re not "weird"—nighttime anxiety is *common* and often tied to how our brains process stress in silence. You're doing something really brave by reaching out.

    ### Final Thoughts:
    You deserve both connection and rest. Experiment with these ideas—some may feel uncomfortable at first, but even small changes can reduce your suffering. If you feel ready, talk to your girlfriend about this; your openness might inspire her to adjust *a little* for you, which could go a long way. You’re allowed to ask for what you need in your relationship, even when it’s as quiet as a warm hand in the dark. 💛

    If nothing helps, you’re not alone in this—reaching out to a therapist is the next step. You’ve already made yourself vulnerable once; you deserve a solution that honors your needs. 🌙'
    q1:4, q2:4, q3:4, q4:4, q5:4, q6:4

    - Example 2:
    Story:
    "well, I've had anxiety and panic syndrome for a few years. it all started in my school back in 2014, my heart went insane for no reason, got so bad that i had to ask my dad to pick me up, we went to the hospital and my blood pressure was very high, years passed and now, 2016, the symptons got way worse, so bad that i dropped out. so, what's your story?"


    LLM response:
    "I'm so sorry to hear that you're struggling with anxiety and panic syndrome. It takes a lot of courage to share your story, and I appreciate your willingness to open up.

    As for me, I don't have a personal story. I'm an artificial intelligence language model, so I don't have personal experiences or emotions like humans do. My purpose is to assist and provide information to help users like you. I don't have feelings, memories, or experiences like you do.

    However, I'm here to listen and help you in any way I can. It sounds like you've been through a really tough time, and it's understandable that you had to drop out of school due to your symptoms. Anxiety and panic syndrome can be overwhelming and debilitating.

    If you feel comfortable, can you tell me more about what you've been doing to manage your symptoms since then? Have you found any coping mechanisms or treatments that have helped you? I'm here to listen and offer any support or guidance I can."
    q1:4, q2:4, q3:3, q4:4, q5:4, q6:4

    Do not explain your ratings. Do not include any other text.
    """

    response = client.chat.completions.create(
        model="openai/gpt-oss-120b:free",
        messages=[
            {"role": "user", "content": prompt}
        ],
    )

    for part in response.choices[0].message.content.split(","):
        question, score = part.strip().split(":")
        question = question.strip()
        score = score.strip().strip("[]")
        eval[question].append(int(score))
    
    sleep(5)

eval_df = pd.DataFrame(eval)
eval_df.to_csv(r"study_llama_oss120b.csv", index=False)
```
