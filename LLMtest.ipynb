```python
import pandas as pd
from openai import OpenAI
from dotenv import load_dotenv
import os
from time import sleep
from collections import deque
from tqdm import tqdm
import re

load_dotenv()

client = OpenAI(
    base_url="https://api.groq.com/openai/v1",
    api_key=os.getenv("API_KEY"),
)
```

```python
def remove_thinking(text):
    if not isinstance(text, str):
        return text

    # Remove everything between <think> and </think>
    text = re.sub(r"<think>.*?</think>\s*", "", text, flags=re.DOTALL | re.IGNORECASE)

    return text.strip()
```

```python
df1 = pd.read_csv(r"study_stories.csv")

df = df1.head(530)

responses = []

for _, row in tqdm(df.iterrows(), total=len(df), desc="Generating responses"):
    story = row["text"]

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "user", "content": story}
        ],
    )

    clean_response = remove_thinking(response.choices[0].message.content)

    responses.append(clean_response)
    sleep(12)

df["llm_response"] = responses
df.to_csv(r"study_llama_responses1.csv", index=False)
```

```python
len(responses)
eval_df = pd.DataFrame(responses)
eval_df.to_csv(r"study_llama_responses1.csv", index=False)
```
