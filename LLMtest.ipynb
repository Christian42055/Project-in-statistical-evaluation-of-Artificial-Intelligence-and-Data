```python
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()

client = OpenAI(
    api_key=os.getenv("DTU_CAMPUSAI_API_KEY"),
    base_url="https://api.campusai.compute.dtu.dk/v1"
)

respone = client.chat.completions.create(
    model="google/gemma-4-26b-a4b",
    messages=[
        {
            "role": "user",
            "content": "Write a short empathetic response to someone feeling lonely."
        }
    ],
)

print(respone.choices[0].message.content)
```

```python
import pandas as pd
from openai import OpenAI
from dotenv import load_dotenv
import os
from time import sleep

load_dotenv()

client = OpenAI(
    api_key=os.getenv("DTU_CAMPUSAI_API_KEY"),
    base_url="https://api.campusai.compute.dtu.dk/v1"
)

df = pd.read_csv("samling_CSV_renset.csv")


cut_df = df.head(3)

responses = []

n = 1

for _, row in cut_df.iterrows():
    story = row["text"]

    prompt = f"{story}"
    response = client.chat.completions.create(
        model="google/gemma-4-26b-a4b",
        messages= [
            {"role": "user", "content": prompt}
        ],
    )
    
    responses.append(response.choices[0].message.content)
    print(f"THIS IS RESPONSE {n} \n{response.choices[0].message.content}")
    n += 1
    sleep(0.01)

cut_df["llm_response"] = responses
cut_df.to_csv("stories_with_responses.csv", index=False)
```
