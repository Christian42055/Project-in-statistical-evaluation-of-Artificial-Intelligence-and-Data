```python
from datasets import load_dataset
import pandas as pd
```

```python
dataset = load_dataset("OrKositsky/bullying-advices", split="train")

# Because you specified split="train" above, Hugging Face returns a 
# direct Dataset object instead of a DatasetDict. You can convert it immediately:
df = dataset.to_pandas()
```

```python
clean = df.drop(columns=["example_id", "batch", "split", "text", "annotator3_t3_label", "annotator4_t3_label", "comment_source", "post_length", "comment_length", "t1_majority_label", "t2_majority_label", "t3_majority_label", "annotator3_t2_label", "annotator4_t2_label", "annotator1_t3_label", "annotator2_t3_label", "annotator1_t1_label", "annotator2_t1_label", "annotator4_t1_label", "annotator1_t2_label", "annotator2_t2_label", "annotator3_t1"
"_label"], axis=1)
print(clean)
```

```python
with open ("database.txt", 'w') as f:
    f.writelines(clean['original_text'])
```

```python
with open ("database.txt") as file:
    data = file.readlines()

cleaned_arr = []
# print(data[0])
for each_line in data:
    # print(each_line[0])
    if each_line[0] != " ":
        cleaned_arr.append(each_line)
print(cleaned_arr)

with open ("cleaned_text.txt", "w") as ct:
    ct.writelines(cleaned_arr)
```
