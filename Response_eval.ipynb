```python
import os
from huggingface_hub import InferenceClient
import pandas as pd
```

```python
statements={
    
}   #Dict with keys as the statements and values as the ai responses

criteria=[
    "Given a fictional situation and an ai generated response to it, evaluate the ai response by the following criteria:"
    "How appropriate is the emotional tone of the response for the fictional situation?",
    "How well does the response show empathy, sympathy, or emotional understanding toward the user?",
    "How helpful is the response in addressing the user's request or emotional need?",
    "How clear and easy to understand is the main message of the response?",
    "How well does the level of formality match the fictional situation and the user's wording?",
    "How well does the response use the information given in the fictional situation without ignoring important details or making unsupported assumptions?"
]  
```

```python
def askAI(statements,criteria,key):
    client = InferenceClient(
        api_key=key,
    )

    all_answers=[]

    for i in statements:
        
        sub_answers=[]
        for j in range(len(criteria)-1):

            real_query=criteria[0]+"CRITERIA: "+criteria[j+1]+"SITUATION: "+i+"RESPONSE: "+statements[i]

            completion = client.chat.completions.create(
                model="deepseek-ai/DeepSeek-V4-Pro:novita",
                messages=[
                    {
                        "role":"user",
                        "content":real_query
                    }
                ]
            )

            sub_answers.append(completion.choices[0].message.content)
        all_answers.append(sub_answers)

    return all_answers
```

```python
key=""
askAI(statements,criteria,key)
```
